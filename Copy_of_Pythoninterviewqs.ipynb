### Question 1: Reverse a String

**Problem Statement:** Write a Python function that takes a string as input and returns the string reversed.

**Solution Code:**
```python
def reverse_string(s):
    return s[::-1]
```

**Sample Input/Output:**
```
Input: "hello"
Output: "olleh"

Input: "Python"
Output: "nohtyP"
```

**Explanation:**
This solution uses Python's slicing feature. `[::-1]` creates a reversed copy of the string. It's a concise and Pythonic way to reverse a string.

**Time Complexity:** O(n), where 'n' is the length of the string. Although it's a single operation, Python internally creates a new string by iterating through the original string, making it proportional to the string's length.

**Space Complexity:** O(n), where 'n' is the length of the string. A new string is created to store the reversed version.

### Question 2: Find the Factorial of a Number

**Problem Statement:** Write a Python function to calculate the factorial of a non-negative integer.

**Solution Code (Iterative):**
```python
def factorial_iterative(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result
```

**Solution Code (Recursive):**
```python
def factorial_recursive(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        return n * factorial_recursive(n - 1)
```

**Sample Input/Output:**
```
Input: 5
Output: 120 (5 * 4 * 3 * 2 * 1)

Input: 0
Output: 1
```

**Explanation:**
- **Iterative:** This approach uses a loop to multiply numbers from 1 to `n`. It's generally preferred for factorials as it avoids potential stack overflow issues with large `n` that recursion might face.
- **Recursive:** This approach defines the factorial of `n` as `n` times the factorial of `n-1`, with a base case of `factorial(0) = 1`. It's elegant but can be less efficient due to function call overhead.

**Time Complexity:** O(n) for both iterative and recursive approaches, as they both perform 'n' multiplications.

**Space Complexity:**
- **Iterative:** O(1), as it uses a constant amount of extra space.
- **Recursive:** O(n) due to the call stack depth, as 'n' function calls are made and stored on the stack.

### Question 3: Check if a Number is Prime

**Problem Statement:** Write a Python function that takes an integer as input and returns `True` if it's a prime number, `False` otherwise.

**Solution Code:**
```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
```

**Sample Input/Output:**
```
Input: 7
Output: True

Input: 10
Output: False (10 is divisible by 2 and 5)

Input: 1
Output: False
```

**Explanation:**
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

1.  Numbers less than or equal to 1 are not prime.
2.  We only need to check for divisors up to the square root of `num`. If `num` has a divisor `d > sqrt(num)`, then it must also have a divisor `num/d < sqrt(num)`. So, checking up to `sqrt(num)` is sufficient.
3.  If any number in the range `[2, sqrt(num)]` divides `num` evenly, `num` is not prime.

**Time Complexity:** O(sqrt(n)), where 'n' is the input number. This is because the loop runs up to the square root of 'n'.

**Space Complexity:** O(1), as it uses a constant amount of extra space.

### Question 4: Check for Palindrome

**Problem Statement:** Write a Python function that takes a string or a number as input and returns `True` if it's a palindrome, `False` otherwise. A palindrome reads the same forwards and backward.

**Solution Code (String):**
```python
def is_palindrome_string(s):
    # Convert to lowercase and remove non-alphanumeric characters for robust check
    s = ''.join(char.lower() for char in s if char.isalnum())
    return s == s[::-1]
```

**Solution Code (Number):**
```python
def is_palindrome_number(n):
    if n < 0: # Negative numbers cannot be palindromes
        return False
    return str(n) == str(n)[::-1]
```

**Sample Input/Output:**
```
Input: "madam"
Output: True

Input: "A man, a plan, a canal: Panama"
Output: True

Input: 121
Output: True

Input: 123
Output: False
```

**Explanation:**
- **String Palindrome:** The most common approach for strings is to compare the original string with its reversed version. For a more robust check, it's good practice to normalize the string by converting it to lowercase and removing non-alphanumeric characters before comparison.
- **Number Palindrome:** For numbers, convert the number to a string and then apply the same string reversal logic.

**Time Complexity:** O(n), where 'n' is the length of the string (or number of digits for a number). String slicing takes linear time.

**Space Complexity:** O(n), as a new string is created for the reversed version and potentially for the normalized version.

### Question 5: Generate Fibonacci Sequence

**Problem Statement:** Write a Python function that generates the first 'n' numbers of the Fibonacci sequence. The Fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones, usually starting with 0 and 1. (e.g., 0, 1, 1, 2, 3, 5, 8, ...)

**Solution Code (Iterative):**
```python
def fibonacci_sequence_iterative(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    else:
        fib_list = [0, 1]
        while len(fib_list) < n:
            next_fib = fib_list[-1] + fib_list[-2]
            fib_list.append(next_fib)
        return fib_list
```

**Solution Code (Recursive - less efficient for large 'n'):**
```python
def fibonacci_recursive_single(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_recursive_single(n-1) + fibonacci_recursive_single(n-2)

def fibonacci_sequence_recursive(n):
    if n <= 0:
        return []
    return [fibonacci_recursive_single(i) for i in range(n)]
```

**Sample Input/Output:**
```
Input: 5
Output: [0, 1, 1, 2, 3]

Input: 8
Output: [0, 1, 1, 2, 3, 5, 8, 13]

Input: 1
Output: [0]
```

**Explanation:**
- **Iterative:** This is the most efficient way to generate the sequence. It stores the previous two numbers and iteratively calculates the next one, appending it to a list.
- **Recursive:** While elegant, the naive recursive solution (calculating `fibonacci_recursive_single` for each `n`) has overlapping subproblems, leading to repeated calculations. For generating a sequence, it's generally much slower than the iterative method for larger 'n'.

**Time Complexity:**
- **Iterative:** O(n), as it calculates each Fibonacci number up to 'n' once.
- **Recursive (Naive):** O(2^n) due to redundant calculations. Can be optimized to O(n) using memoization or dynamic programming, but the provided naive recursive function is exponential.

**Space Complexity:** O(n) for both approaches (when storing the entire sequence), as a list of 'n' numbers is returned.

### Question 6: Find the Largest Element in a List

**Problem Statement:** Write a Python function that takes a list of numbers as input and returns the largest element in the list.

**Solution Code:**
```python
def find_largest_element(numbers):
    if not numbers: # Handle empty list case
        return None # Or raise an error, depending on requirements
    
    max_element = numbers[0]
    for num in numbers:
        if num > max_element:
            max_element = num
    return max_element
```

**Alternative Solution (using `max()` function):**
```python
def find_largest_element_builtin(numbers):
    if not numbers:
        return None
    return max(numbers)
```

**Sample Input/Output:**
```
Input: [3, 1, 4, 1, 5, 9, 2, 6]
Output: 9

Input: [-10, -5, -20]
Output: -5

Input: []
Output: None
```

**Explanation:**
- The manual solution initializes `max_element` with the first element and then iterates through the rest of the list, updating `max_element` whenever a larger number is found. It handles the edge case of an empty list.
- Python's built-in `max()` function provides a more concise and often optimized way to achieve the same result. For interviews, it's often good to know both the manual implementation and the built-in alternatives.

**Time Complexity:** O(n), where 'n' is the number of elements in the list. The function iterates through the list once.

**Space Complexity:** O(1), as it uses a constant amount of extra space (a few variables to store `max_element` and loop counter).

### Question 7: Count Vowels in a String

**Problem Statement:** Write a Python function that takes a string as input and returns the number of vowels (a, e, i, o, u, case-insensitive) in the string.

**Solution Code:**
```python
def count_vowels(s):
    vowels = "aeiou"
    count = 0
    for char in s:
        if char.lower() in vowels:
            count += 1
    return count
```

**Sample Input/Output:**
```
Input: "Hello World"
Output: 3 (e, o, o)

Input: "Python Programming"
Output: 5 (o, o, a, i, o)

Input: "Rhythm"
Output: 0
```

**Explanation:**
1.  Define a set or string of vowels for quick lookup.
2.  Initialize a counter to 0.
3.  Iterate through each character of the input string.
4.  Convert each character to lowercase to handle case-insensitivity.
5.  If the character is in the set of vowels, increment the counter.
6.  Return the final count.

**Time Complexity:** O(n), where 'n' is the length of the input string. The function iterates through the string once.

**Space Complexity:** O(1), as the `vowels` string/set and `count` variable use a constant amount of extra space, regardless of the input string's length.

### Question 8: Remove Duplicates from a List

**Problem Statement:** Write a Python function that takes a list as input and returns a new list with all duplicate elements removed, preserving the original order of the first occurrence.

**Solution Code:**
```python
def remove_duplicates(input_list):
    seen = set() # Use a set for efficient lookup
    result = []
    for item in input_list:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
```

**Alternative Solution (if order doesn't matter):**
```python
def remove_duplicates_unordered(input_list):
    return list(set(input_list))
```

**Sample Input/Output:**
```
Input: [1, 2, 2, 3, 4, 3, 5]
Output: [1, 2, 3, 4, 5]

Input: ['apple', 'banana', 'apple', 'orange', 'banana']
Output: ['apple', 'banana', 'orange']

Input: []
Output: []
```

**Explanation:**
- **Preserving Order:** The first solution uses a `set` to keep track of elements already `seen`. For each item in the input list, it checks if the item is in `seen`. If not, it adds the item to `seen` and appends it to the `result` list. This maintains the order of the first occurrence.
- **Order Doesn't Matter:** If the order of elements in the output list is not important, converting the list to a `set` and then back to a `list` is the most concise way to remove duplicates. Sets inherently store only unique elements.

**Time Complexity:** O(n), where 'n' is the number of elements in the list. Using a `set` for lookup makes the `item not in seen` check average O(1).

**Space Complexity:** O(n), as both the `seen` set and `result` list can store up to 'n' unique elements in the worst case (all elements are unique).

### Question 9: Check if Two Strings are Anagrams

**Problem Statement:** Write a Python function that takes two strings as input and returns `True` if they are anagrams of each other, `False` otherwise. Anagrams are words or phrases formed by rearranging the letters of another, using all the original letters exactly once.

**Solution Code:**
```python
from collections import Counter

def are_anagrams(s1, s2):
    # Anagrams must have the same length
    if len(s1) != len(s2):
        return False
    
    # Normalize strings (e.g., convert to lowercase, remove spaces) if needed
    # For this problem, we'll assume basic strings without spaces/special chars
    # s1 = s1.lower().replace(' ', '')
    # s2 = s2.lower().replace(' ', '')

    return Counter(s1) == Counter(s2)
```

**Alternative Solution (using sorted strings):**
```python
def are_anagrams_sorted(s1, s2):
    if len(s1) != len(s2):
        return False
    return sorted(s1) == sorted(s2)
```

**Sample Input/Output:**
```
Input: s1 = "listen", s2 = "silent"
Output: True

Input: s1 = "hello", s2 = "world"
Output: False

Input: s1 = "Debit Card", s2 = "Bad Credit"
Output: True (if normalization for spaces/case is applied, but not in current code)
```

**Explanation:**
- **Using `collections.Counter`:** This is a very Pythonic and efficient way. `Counter` creates a dictionary-like object where keys are characters and values are their frequencies. If two strings are anagrams, their character counts must be identical.
- **Using `sorted()`:** If two strings are anagrams, their sorted versions will be identical. This approach sorts both strings and compares them.

**Time Complexity:**
- **`Counter` method:** O(n), where 'n' is the length of the strings. Creating counters involves iterating through strings once. Dictionary comparisons are also efficient.
- **`sorted()` method:** O(n log n), due to the sorting operation.

**Space Complexity:**

### Question 10: Find the Sum of Digits of a Number

**Problem Statement:** Write a Python function that takes a non-negative integer as input and returns the sum of its digits.

**Solution Code (Iterative):**
```python
def sum_digits_iterative(n):
    if n < 0:
        raise ValueError("Input must be a non-negative integer")
    
    total_sum = 0
    while n > 0:
        digit = n % 10  # Get the last digit
        total_sum += digit
        n //= 10      # Remove the last digit
    return total_sum
```

**Alternative Solution (using string conversion):**
```python
def sum_digits_string(n):
    if n < 0:
        raise ValueError("Input must be a non-negative integer")
    
    return sum(int(digit) for digit in str(n))
```

**Sample Input/Output:**
```
Input: 123
Output: 6 (1 + 2 + 3)

Input: 4567
Output: 22 (4 + 5 + 6 + 7)

Input: 0
Output: 0
```

**Explanation:**
- **Iterative (Mathematical):** This method uses the modulo operator (`%`) to extract the last digit and integer division (`//`) to remove it, repeating until the number becomes 0. It's a classic algorithm for digit manipulation.
- **String Conversion:** This approach converts the number to a string, iterates through each character (which is a digit), converts it back to an integer, and sums them up. This is often more concise in Python.

**Time Complexity:** O(log N), where 'N' is the input number. The number of iterations or digits is proportional to the number of digits in N, which is `log10(N)`.

**Space Complexity:**
- **Iterative:** O(1), as it uses a constant amount of extra space.
- **String Conversion:** O(log N) for converting the number to a string.

### Question 11: Find Missing Number in a Sequence

**Problem Statement:** Given a list of 'n' distinct numbers in the range `[0, n]`, return the only number in the range that is missing from the list.

**Solution Code (Using Summation):**
```python
def find_missing_number_sum(nums):
    n = len(nums)
    # Sum of numbers from 0 to n is n * (n + 1) / 2
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum
```

**Solution Code (Using XOR):**
```python
def find_missing_number_xor(nums):
    n = len(nums)
    # XOR all numbers from 0 to n
    xor_sum = 0
    for i in range(n + 1):
        xor_sum ^= i
    
    # XOR all numbers in the given list
    for num in nums:
        xor_sum ^= num
        
    return xor_sum
```

**Sample Input/Output:**
```
Input: [3, 0, 1]
Output: 2 (n=3, range [0, 3], numbers are 0, 1, 3. Missing 2)

Input: [9, 6, 4, 2, 3, 5, 7, 0, 1]
Output: 8 (n=9, range [0, 9]. Missing 8)

Input: [0, 1]
Output: 2 (n=2, range [0, 2]. Missing 2)
```

**Explanation:**
- **Summation Method:** This approach utilizes the mathematical property that the sum of an arithmetic progression `0, 1, ..., n` is `n * (n + 1) / 2`. By calculating the expected sum and subtracting the actual sum of the given numbers, the difference will be the missing number.
- **XOR Method:** This is an elegant bit manipulation solution. The XOR operation (`^`) has the property `a ^ a = 0` and `a ^ 0 = a`. If we XOR all numbers from `0` to `n` with all numbers present in the list, all numbers that appear twice (once in the `0..n` sequence and once in the list) will cancel each other out (XOR to 0). The final result will be the single missing number.

**Time Complexity:** O(n) for both methods, as they iterate through the list of numbers once or twice.

**Space Complexity:** O(1) for both methods, as they use a constant amount of extra space.

### Question 12: Implement a Basic Calculator (Addition, Subtraction)

**Problem Statement:** Write a Python function that takes a string representing a simple arithmetic expression (e.g., "10 + 5", "20 - 7") with only addition and subtraction of positive integers, and returns the result. Assume expressions are always valid and well-formatted with spaces.

**Solution Code:**
```python
def basic_calculator(expression):
    parts = expression.split()
    
    # Initialize result with the first number
    result = int(parts[0])
    
    # Iterate through the rest of the parts, taking operator and operand
    for i in range(1, len(parts), 2):
        operator = parts[i]
        operand = int(parts[i+1])
        
        if operator == '+':
            result += operand
        elif operator == '-':
            result -= operand
        # Could add more operators here
            
    return result
```

**Sample Input/Output:**
```
Input: "10 + 5"
Output: 15

Input: "20 - 7"
Output: 13

Input: "100 - 50 + 20"
Output: 70

Input: "5"
Output: 5
```

**Explanation:**
1.  Split the input `expression` string by spaces to get individual components (numbers and operators).
2.  Initialize the `result` with the first number in the expression.
3.  Iterate through the remaining parts of the expression, stepping by two (`operator`, `operand`).
4.  Based on the `operator` ('+' or '-'), perform the corresponding arithmetic operation on the `result` and the `operand`.
5.  Return the final `result`.

**Time Complexity:** O(L), where 'L' is the length of the input expression string (dominated by `split()` and loop iterations).

**Space Complexity:** O(L), for storing the `parts` list after splitting the expression string.

### Question 13: Find the Intersection of Two Lists

**Problem Statement:** Write a Python function that takes two lists as input and returns a new list containing elements that are common to both lists. The order of elements in the output list does not matter.

**Solution Code (Using Sets):**
```python
def list_intersection(list1, list2):
    set1 = set(list1)
    set2 = set(list2)
    return list(set1.intersection(set2))
```

**Alternative Solution (Iterative):**
```python
def list_intersection_iterative(list1, list2):
    intersection_list = []
    # Iterate through the smaller list for efficiency
    if len(list1) > len(list2):
        list1, list2 = list2, list1 # Swap to ensure list1 is smaller

    for item in list1:
        if item in list2 and item not in intersection_list:
            intersection_list.append(item)
    return intersection_list
```

**Sample Input/Output:**
```
Input: list1 = [1, 2, 3, 4], list2 = [3, 4, 5, 6]
Output: [3, 4] (order might vary with set method)

Input: list1 = [10, 20, 30], list2 = [30, 40, 10]
Output: [10, 30]

Input: list1 = [1, 2], list2 = [3, 4]
Output: []
```

**Explanation:**
- **Using Sets:** This is the most efficient and Pythonic way. Convert both lists to sets, then use the `intersection()` method to find common elements. Finally, convert the resulting set back to a list.
- **Iterative:** This method iterates through one list and checks if each element exists in the other list. To prevent duplicates in the output, it also checks if the element is already in `intersection_list`. Iterating through the smaller list first can offer minor performance benefits.

**Time Complexity:**
- **Using Sets:** O(m + n), where 'm' and 'n' are the lengths of the two lists. Converting to set takes O(length), and set intersection is efficient.
- **Iterative:** O(m * n) in the worst case, if `item in list2` takes O(n) for each item in list1 (if `list2` is not sorted or a hash set). It can be optimized to O(m log n) or O(m + n) if `list2` is sorted or converted to a set respectively.

**Space Complexity:**
- **Using Sets:** O(m + n) in the worst case, for storing two sets.
- **Iterative:** O(min(m, n)) for the `intersection_list`.

### Question 16: Sort a List of Strings by Length

**Problem Statement:** Write a Python function that takes a list of strings as input and returns a new list sorted by the length of the strings, from shortest to longest. If two strings have the same length, their relative order should be preserved (stable sort).

**Solution Code:**
```python
def sort_strings_by_length(string_list):
    return sorted(string_list, key=len)
```

**Sample Input/Output:**
```
Input: ["apple", "banana", "cat", "dog", "elephant"]
Output: ["cat", "dog", "apple", "banana", "elephant"]

Input: ["a", "bb", "ccc", "d"]
Output: ["a", "d", "bb", "ccc"]

Input: ["hi", "hello", "world", "a"]
Output: ["a", "hi", "world", "hello"]
```

**Explanation:**
- Python's built-in `sorted()` function (or the `list.sort()` method) accepts an optional `key` argument. The `key` argument should be a function that is called on each element of the list to produce a value for comparison. In this case, `key=len` tells `sorted()` to use the length of each string as the basis for sorting. Python's sorting algorithms are stable, meaning elements with equal keys maintain their original relative order.

**Time Complexity:** O(N log N), where 'N' is the number of strings in the list. This is the typical time complexity for efficient comparison-based sorting algorithms.

**Space Complexity:** O(N) for `sorted()` as it returns a new list. `list.sort()` performs an in-place sort and typically has O(log N) or O(N) space complexity depending on the implementation.

### Question 17: Find the Most Frequent Element in a List

**Problem Statement:** Write a Python function that takes a list as input and returns the element that appears most frequently in the list. If there are multiple elements with the same highest frequency, return any one of them.

**Solution Code (Using `collections.Counter`):**
```python
from collections import Counter

def most_frequent_element(input_list):
    if not input_list:
        return None

    counts = Counter(input_list)
    # Counter.most_common(1) returns a list of (element, count) tuples
    # We want the element of the first tuple
    return counts.most_common(1)[0][0]
```

**Alternative Solution (Manual Counting):**
```python
def most_frequent_element_manual(input_list):
    if not input_list:
        return None

    counts = {}
    for item in input_list:
        counts[item] = counts.get(item, 0) + 1

    max_count = 0
    most_freq_item = None
    for item, count in counts.items():
        if count > max_count:
            max_count = count
            most_freq_item = item
    
    return most_freq_item
```

**Sample Input/Output:**
```
Input: [1, 3, 2, 1, 3, 3, 1]
Output: 1 (or 3, depending on tie-breaking)

Input: ['a', 'b', 'a', 'c', 'a', 'b']
Output: 'a'

Input: []
Output: None
```

**Explanation:**
- **Using `collections.Counter`:** This is the most concise and efficient Pythonic approach. `Counter` is designed for counting hashable objects. `counts.most_common(1)` returns a list of the single most common element and its count. We extract the element from this result.
- **Manual Counting:** This involves creating a dictionary to store the frequency of each element. After counting all elements, iterate through the dictionary to find the element with the maximum count.

**Time Complexity:** O(n), where 'n' is the number of elements in the list. Both methods involve iterating through the list once to count frequencies.

**Space Complexity:** O(k), where 'k' is the number of unique elements in the list (at most 'n'). This space is used to store the frequency map/Counter object.

**Problem Statement:** Write a Python function that takes a string containing just the characters '(', ')', '{', '}', '[' and ']', and determines if the input string is valid. An input string is valid if:
1.  Open brackets must be closed by the same type of brackets.
2.  Open brackets must be closed in the correct order.
3.  Every close bracket has a corresponding open bracket of the same type.

**Solution Code:**
```python
def is_valid_parentheses(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in mapping.values():  # It's an opening bracket
            stack.append(char)
        elif char in mapping.keys():  # It's a closing bracket
            if not stack or stack.pop() != mapping[char]:
                return False
        # Ignore other characters if any, or raise error if input must be only brackets

    return not stack  # True if stack is empty (all brackets matched)
```

**Sample Input/Output:**
```
Input: "()[]{}"
Output: True

Input: "([{}])"
Output: True

Input: "(]"
Output: False

Input: "{[]}"
Output: True

Input: "((("
Output: False
```

**Explanation:**
This problem is a classic application of a stack data structure.
1.  Initialize an empty `stack` to store opening brackets.
2.  Create a `mapping` of closing brackets to their corresponding opening brackets.
3.  Iterate through the input string:
    -   If an opening bracket is encountered, push it onto the stack.
    -   If a closing bracket is encountered:
        -   Check if the stack is empty (meaning no corresponding open bracket). If so, it's invalid.
        -   Pop the top element from the stack and compare it with the required opening bracket (from the `mapping`). If they don't match, it's invalid.
4.  After iterating through the entire string, if the stack is empty, all brackets were matched correctly, and the string is valid. Otherwise, there are unmatched opening brackets, and the string is invalid.

**Time Complexity:** O(n), where 'n' is the length of the string. Each character is processed once (pushed or popped from the stack).

**Space Complexity:** O(n) in the worst case, if all characters are opening brackets, the stack could store all 'n' characters.

### Question 20: Merge Two Sorted Lists

**Problem Statement:** Write a Python function that takes two sorted lists of integers as input and merges them into a single sorted list. You may assume that both lists are sorted in ascending order.

**Solution Code (Iterative):**
```python
def merge_sorted_lists(list1, list2):
    merged_list = []
    i = 0  # Pointer for list1
    j = 0  # Pointer for list2

    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1
    
    # Append any remaining elements from list1 (if list2 was exhausted)
    while i < len(list1):
        merged_list.append(list1[i])
        i += 1
    
    # Append any remaining elements from list2 (if list1 was exhausted)
    while j < len(list2):
        merged_list.append(list2[j])
        j += 1
        
    return merged_list
```

**Explanation:**

### Question 21: Implement Binary Search

**Problem Statement:** Write a Python function that implements binary search on a sorted list of integers. The function should take a sorted list (`arr`) and a `target` value as input, and return the index of the `target` if found, otherwise return -1.

**Solution Code (Iterative):**
```python
def binary_search_iterative(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2  # Calculate middle index
        
        if arr[mid] == target:
            return mid  # Target found
        elif arr[mid] < target:
            low = mid + 1 # Target is in the right half
        else: # arr[mid] > target
            high = mid - 1 # Target is in the left half
            
    return -1 # Target not found
```

**Solution Code (Recursive):**
```python
def binary_search_recursive(arr, target, low, high):
    if low > high:
        return -1  # Target not found

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, high)
    else:
        return binary_search_recursive(arr, target, low, mid - 1)

# Wrapper function for initial call
def binary_search_recursive_wrapper(arr, target):
    return binary_search_recursive(arr, target, 0, len(arr) - 1)
```

**Sample Input/Output:**
```
Input: arr = [1, 2, 3, 4, 5, 6, 7, 8, 9], target = 5
Output: 4

Input: arr = [10, 20, 30, 40, 50], target = 25
Output: -1

Input: arr = [1, 3, 5, 7, 9], target = 1
Output: 0

Input: arr = [], target = 5
Output: -1
```

**Explanation:**
Binary search is an efficient algorithm for finding an item from a sorted list. It works by repeatedly dividing in half the portion of the list that could contain the item, until you've narrowed down the possible locations to just one.
1.  Initialize `low` to the first index and `high` to the last index of the list.
2.  While `low` is less than or equal to `high`:
    -   Calculate `mid` (the middle index).
    -   If the element at `mid` is the `target`, return `mid`.
    -   If the element at `mid` is less than the `target`, the `target` must be in the right half, so update `low = mid + 1`.
    -   If the element at `mid` is greater than the `target`, the `target` must be in the left half, so update `high = mid - 1`.
3.  If the loop finishes, the `target` was not found, so return -1.

**Time Complexity:** O(log n), where 'n' is the number of elements in the list. This is because the search space is halved in each step.

**Space Complexity:**
- **Iterative:** O(1), as it uses a constant amount of extra space.
- **Recursive:** O(log n) due to the recursive call stack depth.

### Question 22: Implement Bubble Sort

**Problem Statement:** Write a Python function to sort a list of numbers in ascending order using the Bubble Sort algorithm. (For interview purposes, demonstrating understanding of basic sorting algorithms is key, even if not the most efficient).

**Solution Code:**
```python
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Last i elements are already in place, so no need to check them
        for j in range(0, n - i - 1):
            # Traverse the array from 0 to n-i-1
            # Swap if the element found is greater than the next element
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j] # Pythonic swap
    return arr
```

**Sample Input/Output:**
```
Input: [64, 34, 25, 12, 22, 11, 90]
Output: [11, 12, 22, 25, 34, 64, 90]

Input: [5, 1, 4, 2, 8]
Output: [1, 2, 4, 5, 8]

Input: []
Output: []

Input: [7]
Output: [7]
```

**Explanation:**
Bubble Sort is a simple comparison-based sorting algorithm. It repeatedly steps through the list, compares adjacent elements and swaps them if they are in the wrong order. The pass through the list is repeated until no swaps are needed, which indicates that the list is sorted.

- The outer loop runs `n-1` times (for `n` elements, `n-1` passes are needed).
- The inner loop compares adjacent elements and swaps them if the left one is greater than the right one. With each pass of the outer loop, the largest unsorted element "bubbles" to its correct position at the end of the unsorted portion of the list. That's why `n - i - 1` is used for the inner loop range, reducing the comparison range as elements get sorted.

**Time Complexity:** O(n^2) in the worst and average cases. This is because there are two nested loops, each potentially running up to 'n' times.

**Space Complexity:** O(1), as it only requires a constant amount of extra space for temporary variables during swaps and loop counters. It sorts the array in-place.

### Question 26: Implement Selection Sort

**Problem Statement:** Write a Python function to sort a list of numbers in ascending order using the Selection Sort algorithm.

**Solution Code:**
```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # Find the minimum element in remaining unsorted array
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        # Swap the found minimum element with the first element of the unsorted part
        arr[i], arr[min_idx] = arr[min_idx], arr[i] # Pythonic swap
    return arr
```

**Sample Input/Output:**
```
Input: [64, 25, 12, 22, 11]
Output: [11, 12, 22, 25, 64]

Input: [5, 1, 4, 2, 8]
Output: [1, 2, 4, 5, 8]

Input: []
Output: []

Input: [7]
Output: [7]
```

**Explanation:**
Selection sort is an in-place comparison sorting algorithm. It divides the input list into two parts: a sorted sublist at the beginning and an unsorted sublist at the end. The algorithm repeatedly finds the minimum element from the unsorted sublist and swaps it with the leftmost element of the unsorted sublist, expanding the sorted sublist.

1.  The outer loop iterates from the first element to the second-to-last element (`n-1` times).
2.  In each iteration of the outer loop, `min_idx` is initialized to the current `i`. The inner loop then scans the *rest* of the unsorted part of the array (`from i+1 to n-1`) to find the actual minimum element's index.
3.  Once the minimum element in the unsorted portion is found at `min_idx`, it is swapped with the element at the current position `i`.

**Time Complexity:** O(N^2) in all cases (best, average, worst). This is because it always makes N-1 passes, and in each pass, it performs roughly N comparisons to find the minimum element.

**Space Complexity:** O(1), as it sorts the array in-place and uses a constant amount of extra space for variables.

### Question 27: Check if a Linked List Has a Cycle (using Floyd's Cycle-Finding Algorithm)

**Problem Statement:** Given the `head` of a singly linked list, return `True` if there is a cycle in the linked list. Otherwise, return `False`. A cycle means that some node in the list can be reached again by continuously following the `next` pointer.

*(Note: For freshers, this might require understanding of Linked Lists first. If the problem expects a solution without explicit linked list node definition, it usually implies a list of numbers where a value indicates the next index, or an abstraction for a node. For this solution, we define a simple `ListNode` class.)*

**Solution Code:**
```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def has_cycle(head):
    if not head or not head.next:
        return False # No list or single node, no cycle

    slow = head
    fast = head.next

    while fast and fast.next:
        if slow == fast:
            return True # Cycle detected
        slow = slow.next
        fast = fast.next.next

    return False # No cycle found
```

**Sample Input/Output (Conceptual with visual aid):**
```
# Example 1: Cycle Present
# head -> 3 -> 2 -> 0 -> -4
#              ^         |
#              |---------|
# Output: True

# Example 2: No Cycle
# head -> 1 -> 2 -> null
# Output: False

# Example 3: Single node, no cycle
# head -> 1 -> null
# Output: False
```

**Explanation (Floyd's Cycle-Finding Algorithm - Tortoise and Hare):**
This algorithm uses two pointers, `slow` and `fast`.
1.  Initialize `slow` pointer to `head` and `fast` pointer to `head.next` (or both to `head` and fast moves first).
2.  Move `slow` one step at a time (`slow = slow.next`).
3.  Move `fast` two steps at a time (`fast = fast.next.next`).
4.  If there is a cycle, the `fast` pointer will eventually catch up to the `slow` pointer (they will meet at some node). When they meet, return `True`.
5.  If `fast` or `fast.next` becomes `None`, it means the end of the list was reached, and there is no cycle. Return `False`.

**Time Complexity:** O(N), where 'N' is the number of nodes in the linked list. In the worst case (no cycle or cycle at the very end), the fast pointer traverses the list twice.

**Space Complexity:** O(1), as it uses only a constant amount of extra space for the two pointers.

### Question 29: Implement Insertion Sort

**Problem Statement:** Write a Python function to sort a list of numbers in ascending order using the Insertion Sort algorithm.

**Solution Code:**
```python
def insertion_sort(arr):
    # Traverse through 1 to len(arr)
    for i in range(1, len(arr)):
        key = arr[i]
        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key # Place key at its correct position
    return arr
```

**Sample Input/Output:**
```
Input: [12, 11, 13, 5, 6]
Output: [5, 6, 11, 12, 13]

Input: [5, 1, 4, 2, 8]
Output: [1, 2, 4, 5, 8]

Input: []
Output: []

Input: [7]
Output: [7]
```

**Explanation:**
Insertion sort builds the final sorted array (or list) one item at a time. It iterates through the input list and for each element, places it into its correct position within the sorted part of the list.

1.  The outer loop starts from the second element (`i = 1`) because the first element `arr[0]` is considered already sorted.
2.  `key` stores the current element to be inserted into the sorted sub-array.
3.  The inner `while` loop shifts elements in the sorted portion (`arr[0...i-1]`) that are greater than `key` one position to the right to make space for `key`.
4.  Finally, `key` is placed at its correct sorted position.

**Time Complexity:**
-   **Worst Case:** O(N^2) (e.g., when the list is sorted in reverse order).
-   **Average Case:** O(N^2).
-   **Best Case:** O(N) (when the list is already sorted, as the inner `while` loop won't run much).

**Space Complexity:** O(1), as it sorts the array in-place and uses a constant amount of extra space.

**Problem Statement:** Implement a `Queue` data structure using two stacks. A queue follows the First-In, First-Out (FIFO) principle. Implement the `enqueue` (add element to queue) and `dequeue` (remove element from queue) operations.

**Solution Code:**
```python
class QueueUsingTwoStacks:
    def __init__(self):
        self.stack_in = []  # Used for enqueuing elements
        self.stack_out = [] # Used for dequeuing elements

    def enqueue(self, item):
        self.stack_in.append(item)
        print(f"Enqueued: {item}, Queue: {self.stack_in + list(reversed(self.stack_out))}")

    def dequeue(self):
        if not self.stack_out: # If stack_out is empty, transfer from stack_in
            while self.stack_in:
                self.stack_out.append(self.stack_in.pop())
        
        if self.stack_out: # If stack_out is still not empty (after potential transfer)
            dequeued_item = self.stack_out.pop()
            print(f"Dequeued: {dequeued_item}, Queue: {self.stack_in + list(reversed(self.stack_out))}")
            return dequeued_item
        else:
            print("Queue is empty, cannot dequeue.")
            return None

    def peek(self):
        if not self.stack_out:
            while self.stack_in:
                self.stack_out.append(self.stack_in.pop())
        
        if self.stack_out:
            return self.stack_out[-1]
        else:
            return None

    def is_empty(self):
        return not self.stack_in and not self.stack_out

    def size(self):
        return len(self.stack_in) + len(self.stack_out)

# Demo
my_queue = QueueUsingTwoStacks()
my_queue.enqueue(1)
my_queue.enqueue(2)
my_queue.enqueue(3)
print(f"Is queue empty? {my_queue.is_empty()}")
print(f"Queue size: {my_queue.size()}")
my_queue.dequeue() # Should dequeue 1
my_queue.enqueue(4)
my_queue.dequeue() # Should dequeue 2
my_queue.dequeue() # Should dequeue 3
my_queue.dequeue() # Should dequeue 4
my_queue.dequeue() # Should be empty
print(f"Is queue empty? {my_queue.is_empty()}")
```

**Explanation:**
Implementing a queue using two stacks is a common interview question that tests understanding of stack and queue principles. The main idea is:

-   **`enqueue(item)`:** Simply push the `item` onto `stack_in`. This is an O(1) operation.
-   **`dequeue()`:**
    1.  If `stack_out` is not empty, pop and return an element from `stack_out`. This is O(1).
    2.  If `stack_out` is empty, *then* transfer all elements from `stack_in` to `stack_out`. This reverses the order, so the elements are now in FIFO order in `stack_out`. After the transfer, pop and return an element from `stack_out`. This transfer takes O(N) time for N elements, but the amortized time complexity is O(1) because each element is moved twice (once from `stack_in` to `stack_out`, once popped from `stack_out`) over its lifetime.
-   **`peek()`:** Similar logic to `dequeue` for transferring elements if `stack_out` is empty, then return the top of `stack_out` without popping.

**Time Complexity:**
-   `enqueue`: O(1)
-   `dequeue` (Amortized): O(1). In the worst case (when `stack_out` is empty), it takes O(N) to transfer elements, but this happens infrequently.

**Space Complexity:** O(N), where 'N' is the total number of elements in the queue, as both stacks together store all elements.
