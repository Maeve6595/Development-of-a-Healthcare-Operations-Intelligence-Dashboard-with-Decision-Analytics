# SQL Coding Interview Questions for Freshers

````python
import sqlite3
import pandas as pd

def run_sql_query(db_name, query):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()
    cursor.execute(query)
    column_names = [description[0] for description in cursor.description]
    rows = cursor.fetchall()
    conn.close()
    return pd.DataFrame(rows, columns=column_names)

def setup_database(db_name, setup_queries):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()
    for query in setup_queries:
        cursor.execute(query)
    conn.commit()
    conn.close()

print("Helper functions for SQLite database operations are ready.")
````

## Question 1: Find Employees in a Specific Department

**Difficulty:** Easy

**Problem Statement:** Retrieve the `EmployeeName` and `Department` for all employees who work in the 'Sales' department.

````python
db_file = 'employees_q1.db'
setup_queries_q1 = [
    "DROP TABLE IF EXISTS Employees;",
    "CREATE TABLE Employees (EmployeeID INT, EmployeeName TEXT, Department TEXT, Salary INT);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Department, Salary) VALUES (1, 'Alice', 'Sales', 60000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Department, Salary) VALUES (2, 'Bob', 'Marketing', 75000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Department, Salary) VALUES (3, 'Charlie', 'Sales', 65000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Department, Salary) VALUES (4, 'David', 'HR', 50000);"
]
setup_database(db_file, setup_queries_q1)

print("Sample Table: Employees")
display(run_sql_query(db_file, "SELECT * FROM Employees;"))
````

````python
sql_query_q1 = "SELECT EmployeeName, Department FROM Employees WHERE Department = 'Sales';"
print("SQL Query:")
print(sql_query_q1)
````

````python
print("Query Result:")
display(run_sql_query(db_file, sql_query_q1))
````

**Expected Output:**

| EmployeeName | Department |
|:-------------|:-----------|
| Alice        | Sales      |
| Charlie      | Sales      |

**Explanation:** This query selects `EmployeeName` and `Department` from the `Employees` table and filters the results to include only those where the `Department` is 'Sales'.

## Question 5: List Employees and Their Managers

**Difficulty:** Medium

**Problem Statement:** Given an `Employees` table with `EmployeeID`, `EmployeeName`, and `ManagerID`, list each employee and their respective manager's name. A `ManagerID` of `NULL` means the employee has no manager (is a top-level manager).

````python
db_file = 'employees_q5.db'
setup_queries_q5 = [
    "DROP TABLE IF EXISTS Employees;",
    "CREATE TABLE Employees (EmployeeID INT, EmployeeName TEXT, ManagerID INT);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, ManagerID) VALUES (1, 'John', NULL);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, ManagerID) VALUES (2, 'Jane', 1);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, ManagerID) VALUES (3, 'Doe', 1);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, ManagerID) VALUES (4, 'Peter', 2);"
]
setup_database(db_file, setup_queries_q5)

print("Sample Table: Employees")
display(run_sql_query(db_file, "SELECT * FROM Employees;"))
````

````python
sql_query_q5 = """SELECT
    E.EmployeeName AS Employee,
    M.EmployeeName AS Manager
FROM
    Employees E
LEFT JOIN
    Employees M ON E.ManagerID = M.EmployeeID;"""
print("SQL Query:")
print(sql_query_q5)
````

````python
print("Query Result:")
display(run_sql_query(db_file, sql_query_q5))
````

**Expected Output:**

| Employee | Manager |
|:---------|:--------|
| John     | NULL    |
| Jane     | John    |
| Doe      | John    |
| Peter    | Jane    |

**Explanation:** This query uses a `LEFT JOIN` to join the `Employees` table with itself (a self-join). We alias the first instance as `E` (for Employee) and the second as `M` (for Manager). The join condition `E.ManagerID = M.EmployeeID` matches each employee with their manager. A `LEFT JOIN` ensures that all employees are listed, even if they don't have a manager (their `ManagerID` is `NULL`), in which case the `Manager` column will show `NULL`.

## Question 6: Second Highest Salary

**Difficulty:** Medium

**Problem Statement:** Find the second highest salary from the `Employees` table. If there is no second highest salary (e.g., only one unique salary), return `NULL`.

````python
db_file = 'employees_q6.db'
setup_queries_q6 = [
    "DROP TABLE IF EXISTS Employees;",
    "CREATE TABLE Employees (EmployeeID INT, EmployeeName TEXT, Salary INT);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Salary) VALUES (1, 'Alice', 60000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Salary) VALUES (2, 'Bob', 75000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Salary) VALUES (3, 'Charlie', 65000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Salary) VALUES (4, 'David', 50000);",
    "INSERT INTO Employees (EmployeeID, EmployeeName, Salary) VALUES (5, 'Eve', 75000);"
]
setup_database(db_file, setup_queries_q6)

print("Sample Table: Employees")
display(run_sql_query(db_file, "SELECT * FROM Employees;"))
````

````python
sql_query_q6 = """SELECT DISTINCT Salary
FROM Employees
ORDER BY Salary DESC
LIMIT 1 OFFSET 1;"""
print("SQL Query:")
print(sql_query_q6)
````

````python
print("Query Result:")
display(run_sql_query(db_file, sql_query_q6))
````

**Expected Output:**

| Salary |
|:-------|
| 65000  |

**Explanation:**
1.  `SELECT DISTINCT Salary`: Selects only unique salary values.
2.  `FROM Employees`: Specifies the table.
3.  `ORDER BY Salary DESC`: Sorts the salaries in descending order (highest first).
4.  `LIMIT 1 OFFSET 1`: This is key. `OFFSET 1` skips the first row (the highest salary), and `LIMIT 1` then retrieves the next row, which is the second highest unique salary.

## Question 9: Customers Who Placed More Than One Order

**Difficulty:** Medium

**Problem Statement:** Find the names of customers who have placed more than one order.

````python
db_file = 'customers_orders_q9.db'
setup_queries_q9 = [
    "DROP TABLE IF EXISTS Customers;",
    "CREATE TABLE Customers (CustomerID INT, CustomerName TEXT);",
    "INSERT INTO Customers (CustomerID, CustomerName) VALUES (101, 'Alice');",
    "INSERT INTO Customers (CustomerID, CustomerName) VALUES (102, 'Bob');",
    "INSERT INTO Customers (CustomerID, CustomerName) VALUES (103, 'Charlie');",
    "DROP TABLE IF EXISTS Orders;",
    "CREATE TABLE Orders (OrderID INT, CustomerID INT, OrderDate DATE);",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (1, 101, '2023-01-10');",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (2, 102, '2023-01-12');",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (3, 101, '2023-01-15');",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (4, 103, '2023-01-18');",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (5, 101, '2023-01-20');",
    "INSERT INTO Orders (OrderID, CustomerID, OrderDate) VALUES (6, 102, '2023-01-22');"
]
setup_database(db_file, setup_queries_q9)

print("Sample Table: Customers")
display(run_sql_query(db_file, "SELECT * FROM Customers;"))
print("\nSample Table: Orders")
display(run_sql_query(db_file, "SELECT * FROM Orders;"))
````

````python
sql_query_q9 = """SELECT
    C.CustomerName
FROM
    Customers C
JOIN
    Orders O ON C.CustomerID = O.CustomerID
GROUP BY
    C.CustomerName
HAVING
    COUNT(O.OrderID) > 1;"""
print("SQL Query:")
print(sql_query_q9)
````

````python
print("Query Result:")
display(run_sql_query(db_file, sql_query_q9))
````

**Expected Output:**

| CustomerName |
|:-------------|
| Alice        |
| Bob          |

**Explanation:**
1.  `JOIN Customers C ON C.CustomerID = O.CustomerID`: Joins `Customers` and `Orders` tables on their common `CustomerID` to link orders to customer names.
2.  `GROUP BY C.CustomerName`: Groups the results by customer name.
3.  `HAVING COUNT(O.OrderID) > 1`: Filters these groups, keeping only those customers who have more than one order (i.e., `COUNT(O.OrderID)` is greater than 1). `HAVING` is used for filtering on aggregated results, similar to how `WHERE` is used for filtering on individual rows.

## Question 10: Delete Duplicate Rows (Keeping One)

**Difficulty:** Medium

**Problem Statement:** Delete duplicate rows from a table, keeping only one instance of each unique row. For this exercise, we'll demonstrate how to identify duplicates and then conceptualize deletion, as direct `DELETE` with `ROWID` can be complex across all SQL variants. We'll show how to select distinct rows.

````python
db_file = 'products_q10.db'
setup_queries_q10 = [
    "DROP TABLE IF EXISTS Products;",
    "CREATE TABLE Products (ProductID INT, ProductName TEXT, Price INT);",
    "INSERT INTO Products (ProductID, ProductName, Price) VALUES (1, 'Apple', 100);",
    "INSERT INTO Products (ProductID, ProductName, Price) VALUES (2, 'Banana', 50);",
    "INSERT INTO Products (ProductID, ProductName, Price) VALUES (1, 'Apple', 100);",
    "INSERT INTO Products (ProductID, ProductName, Price) VALUES (3, 'Orange', 70);",
    "INSERT INTO Products (ProductID, ProductName, Price) VALUES (2, 'Banana', 50);"
]
setup_database(db_file, setup_queries_q10)

print("Sample Table: Products (with duplicates)")
display(run_sql_query(db_file, "SELECT * FROM Products;"))
````

````python
sql_query_q10 = "SELECT DISTINCT ProductID, ProductName, Price FROM Products;"
print("SQL Query (to select distinct rows):")
print(sql_query_q10)
````

````python
print("Query Result (distinct rows):")
display(run_sql_query(db_file, sql_query_q10))
````

**Expected Output (distinct rows):**

| ProductID | ProductName | Price |
|:----------|:------------|:------|
| 1         | Apple       | 100   |
| 2         | Banana      | 50    |
| 3         | Orange      | 70    |

**Explanation:**
To handle duplicates in a table, a common approach is to identify distinct rows. The `SELECT DISTINCT ProductID, ProductName, Price FROM Products` query retrieves only the unique combinations of `ProductID`, `ProductName`, and `Price` from the `Products` table. This effectively shows what the table would look like with duplicates removed.

In a real-world scenario where you need to physically delete duplicates, you might use a subquery with `ROWID` (in SQLite) or window functions like `ROW_NUMBER()` with `PARTITION BY` (in more advanced SQL databases) to identify and delete all but one instance of each duplicate set. For example, in SQLite, you could do:

```sql
DELETE FROM Products
WHERE ROWID NOT IN (
    SELECT MIN(ROWID)
    FROM Products
    GROUP BY ProductID, ProductName, Price
);
```
This would delete all duplicate rows, keeping the one with the smallest `ROWID`.
